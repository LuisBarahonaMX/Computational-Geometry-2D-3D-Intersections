
#Generación de Líneas Aleatorias y Detección de Intersecciones en 2D

##Este código genera un conjunto de líneas aleatorias en un plano 2D y luego detecta y muestra gráficamente las intersecciones entre estas líneas. El usuario puede especificar la cantidad de líneas que se generarán, y cada línea se dibujará entre dos puntos aleatorios en el plano. Además, se calculará el número de intersecciones y se mostrará junto con los puntos de intersección en el gráfico.

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

def draw_lines(lines, intersections):
    plt.figure(figsize=(8, 8))
    for line in lines:
        (x1, y1), (x2, y2) = line
        plt.plot([x1, x2], [y1, y2], 'bo-')
    for intersection in intersections:
        plt.plot(intersection[0], intersection[1], 'ro')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Líneas y Puntos de Intersección')
    plt.grid(True)
    plt.show()

def calculate_intersection(lines):
    intersections = []
    for i, line1 in enumerate(lines):
        for line2 in lines[i+1:]:
            xdiff = (line1[0][0] - line1[1][0], line2[0][0] - line2[1][0])
            ydiff = (line1[0][1] - line1[1][1], line2[0][1] - line2[1][1])

            def det(a, b):
                return a[0] * b[1] - a[1] * b[0]

            div = det(xdiff, ydiff)
            if div == 0:
                continue
            d = (det(*line1), det(*line2))
            x = det(d, xdiff) / div
            y = det(d, ydiff) / div
            if min(line1[0][0], line1[1][0]) <= x <= max(line1[0][0], line1[1][0]) and min(line1[0][1], line1[1][1]) <= y <= max(line1[0][1], line1[1][1]):
                if min(line2[0][0], line2[1][0]) <= x <= max(line2[0][0], line2[1][0]) and min(line2[0][1], line2[1][1]) <= y <= max(line2[0][1], line2[1][1]):
                    intersections.append((x, y))
    return intersections

def calculate_levels_and_discrepancy(lines):
    levels = defaultdict(int)
    for line in lines:
        (x1, y1), (x2, y2) = line  # Desempaquetar las coordenadas de inicio y fin de la línea
        for x, y in line:  # Iterar sobre los puntos de la línea actual, no sobre todas las líneas
            if (x1 < x < x2 and y < y1) or (x1 < x < x2 and y < y2):
                levels[(x, y)] += 1
    discrepancy = sum(levels.values())
    return levels, discrepancy

def main():
    num_lines = int(input("Ingrese el número de líneas aleatorias a generar: "))
    lines = []
    for i in range(num_lines):
        x1, y1, x2, y2 = np.random.rand(4) * 10
        lines.append(((x1, y1), (x2, y2)))

    intersections = calculate_intersection(lines)
    draw_lines(lines, intersections)

    levels, discrepancy = calculate_levels_and_discrepancy(lines)
    print("Niveles:", levels)
    print("Discrepancia:", discrepancy)

if __name__ == "__main__":
    main()
```

# Detección de Intersecciones entre Líneas en 2D

## Descripción:Este código genera un conjunto de líneas aleatorias en un plano 2D y detecta las intersecciones entre estas líneas. El usuario puede especificar el número de líneas que se generarán, y cada línea se dibujará entre dos puntos aleatorios en el plano. Además, se calculará el número de intersecciones y se mostrarán las coordenadas de las intersecciones. Los resultados se guardarán en un archivo CSV.


```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations
import pandas as pd

def generate_lines(num_lines):
    lines = []
    for i in range(num_lines):
        x1, y1 = np.random.rand(2) * 10
        x2, y2 = np.random.rand(2) * 10
        lines.append(((x1, y1), (x2, y2)))
    return lines

def plot_lines(lines):
    plt.figure(figsize=(8, 8))
    for line in lines:
        (x1, y1), (x2, y2) = line
        plt.plot([x1, x2], [y1, y2], marker='o')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Líneas en 2D')
    plt.grid(True)
    plt.show()

def calculate_intersections(lines):
    intersections = []
    for line1, line2 in combinations(lines, 2):
        (x1, y1), (x2, y2) = line1
        (x3, y3), (x4, y4) = line2

        denominator = ((y4 - y3) * (x2 - x1)) - ((x4 - x3) * (y2 - y1))

        if denominator == 0:
            continue

        ua = (((x4 - x3) * (y1 - y3)) - ((y4 - y3) * (x1 - x3))) / denominator
        ub = (((x2 - x1) * (y1 - y3)) - ((y2 - y1) * (x1 - x3))) / denominator

        if 0 <= ua <= 1 and 0 <= ub <= 1:
            x = x1 + ua * (x2 - x1)
            y = y1 + ua * (y2 - y1)
            intersections.append((x, y))
    return intersections

def main():
    num_lines = int(input("Ingrese el número de líneas a generar: "))
    lines = generate_lines(num_lines)
    plot_lines(lines)
    intersections = calculate_intersections(lines)
    print("Número de intersecciones:", len(intersections))
    print("Coordenadas de las intersecciones:", intersections)

    # Crear un DataFrame con los datos de las intersecciones
    df = pd.DataFrame(intersections, columns=['X', 'Y'])
    print("\nDataFrame con las intersecciones:")
    print(df)

    # Guardar el DataFrame como archivo CSV
    df.to_csv('intersecciones.csv', index=False)
    print("\nEl DataFrame se ha guardado como 'intersecciones.csv'")

if __name__ == "__main__":
    main()

```

# Generación y Visualización de Líneas en 3D

## Descripción:Este código genera un conjunto de líneas aleatorias en un espacio tridimensional (3D) y visualiza su representación gráfica utilizando Matplotlib. El usuario puede especificar el número de líneas que se generarán, y cada línea se dibujará entre dos puntos aleatorios en el espacio 3D. Además, se calcula el volumen aproximado del espacio delimitado por las líneas y la cantidad de caras que conforman las líneas generadas.


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def draw_lines_3d(lines):
    fig = plt.figure(figsize=(10, 10))
    ax = fig.add_subplot(111, projection='3d')
    for line in lines:
        (x1, y1, z1), (x2, y2, z2) = line
        ax.plot([x1, x2], [y1, y2], [z1, z2], 'bo-')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Líneas en 3D')
    plt.show()

def calculate_volume_and_faces(lines):
    # Calcular el volumen aproximado usando el método de Monte Carlo
    num_points = 10000
    max_x = max([max(line[0][0], line[1][0]) for line in lines])
    max_y = max([max(line[0][1], line[1][1]) for line in lines])
    max_z = max([max(line[0][2], line[1][2]) for line in lines])
    points_inside = 0
    for _ in range(num_points):
        x, y, z = np.random.rand(3) * [max_x, max_y, max_z]
        for line in lines:
            (x1, y1, z1), (x2, y2, z2) = line
            if (x1 <= x <= x2 or x2 <= x <= x1) and (y1 <= y <= y2 or y2 <= y <= y1) and (z1 <= z <= z2 or z2 <= z <= z1):
                points_inside += 1
                break
    volume = max_x * max_y * max_z * (points_inside / num_points)

    # Calcular la cantidad de caras
    num_faces = len(lines)

    return volume, num_faces

def main():
    num_lines = int(input("Ingrese el número de líneas aleatorias a generar: "))
    lines = []
    for i in range(num_lines):
        x1, y1, z1, x2, y2, z2 = np.random.rand(6) * 10
        lines.append(((x1, y1, z1), (x2, y2, z2)))

    draw_lines_3d(lines)

    volume, num_faces = calculate_volume_and_faces(lines)
    print("Volumen aproximado:", volume)
    print("Cantidad de caras:", num_faces)

if __name__ == "__main__":
    main()
```

# Detección de Intersecciones entre Líneas en 3D

## Descripción:Este código detecta las intersecciones entre un conjunto de líneas aleatorias en un espacio tridimensional (3D). Utiliza el método de determinantes para calcular si dos líneas se intersectan en el espacio 3D y cuenta el número total de intersecciones encontradas. Además, visualiza gráficamente las líneas generadas en el espacio 3D y muestra el volumen aproximado del espacio delimitado por las líneas y la cantidad de caras que conforman las líneas generadas.


```python
def calculate_intersections(lines):
    num_intersections = 0
    for i, line1 in enumerate(lines):
        for line2 in lines[i+1:]:
            xdiff = (line1[0][0] - line1[1][0], line2[0][0] - line2[1][0])
            ydiff = (line1[0][1] - line1[1][1], line2[0][1] - line2[1][1])
            zdiff = (line1[0][2] - line1[1][2], line2[0][2] - line2[1][2])

            def det(a, b):
                return a[0] * b[1] - a[1] * b[0]

            div_xy = det(xdiff, ydiff)
            div_xz = det(xdiff, zdiff)
            div_yz = det(ydiff, zdiff)

            if div_xy == 0 or div_xz == 0 or div_yz == 0:
                continue

            d_xy = (det(*line1[:2]), det(*line2[:2]))
            d_xz = (det(line1[0][::2], line1[1][::2]), det(line2[0][::2], line2[1][::2]))
            d_yz = (det(line1[0][1:], line1[1][1:]), det(line2[0][1:], line2[1][1:]))

            x = det(d_xy, xdiff) / div_xy
            y = det(d_xy, ydiff) / div_xy
            z = det(d_xz, xdiff) / div_xz
            if (min(line1[0][0], line1[1][0]) <= x <= max(line1[0][0], line1[1][0]) and
                min(line1[0][1], line1[1][1]) <= y <= max(line1[0][1], line1[1][1]) and
                min(line2[0][0], line2[1][0]) <= x <= max(line2[0][0], line2[1][0]) and
                min(line2[0][1], line2[1][1]) <= y <= max(line2[0][1], line2[1][1])):
                num_intersections += 1

    return num_intersections

def main():
    num_lines = int(input("Ingrese el número de líneas aleatorias a generar: "))
    lines = []
    for i in range(num_lines):
        x1, y1, z1, x2, y2, z2 = np.random.rand(6) * 10
        lines.append(((x1, y1, z1), (x2, y2, z2)))

    draw_lines_3d(lines)

    volume, num_faces = calculate_volume_and_faces(lines)
    print("Volumen aproximado:", volume)
    print("Cantidad de caras:", num_faces)

    num_intersections = calculate_intersections(lines)
    print("Número de intersecciones entre líneas:", num_intersections)

if __name__ == "__main__":
    main()
```

# Detección de Intersecciones y Distancias entre Líneas en 2D

# Descripción:Este código genera un conjunto de líneas aleatorias en un plano 2D y detecta las intersecciones entre ellas. Además, calcula las distancias entre las líneas que se intersectan y muestra gráficamente las líneas y los puntos de intersección. El usuario puede especificar el número de líneas a generar y la longitud de las líneas. El programa también calcula los niveles y la discrepancia entre las líneas generadas.


```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from itertools import combinations

def draw_lines(lines, intersections):
    plt.figure(figsize=(8, 8))
    for line in lines:
        (x1, y1), (x2, y2) = line
        plt.plot([x1, x2], [y1, y2], 'bo-')
    for intersection in intersections:
        plt.plot(intersection[0], intersection[1], 'ro')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Líneas y Puntos de Intersección')
    plt.grid(True)
    plt.show()

def calculate_intersection(lines):
    intersections = []
    for line1, line2 in combinations(lines, 2):
        xdiff = (line1[0][0] - line1[1][0], line2[0][0] - line2[1][0])
        ydiff = (line1[0][1] - line1[1][1], line2[0][1] - line2[1][1])

        def det(a, b):
            return a[0] * b[1] - a[1] * b[0]

        div = det(xdiff, ydiff)
        if div == 0:
            continue
        d = (det(*line1), det(*line2))
        x = det(d, xdiff) / div
        y = det(d, ydiff) / div
        if min(line1[0][0], line1[1][0]) <= x <= max(line1[0][0], line1[1][0]) and min(line1[0][1], line1[1][1]) <= y <= max(line1[0][1], line1[1][1]):
            if min(line2[0][0], line2[1][0]) <= x <= max(line2[0][0], line2[1][0]) and min(line2[0][1], line2[1][1]) <= y <= max(line2[0][1], line2[1][1]):
                intersections.append((x, y))
    return intersections

def calculate_levels_and_discrepancy(lines):
    levels = defaultdict(int)
    for line in lines:
        (x1, y1), (x2, y2) = line
        for x, y in line:
            if (x1 < x < x2 and y < y1) or (x1 < x < x2 and y < y2):
                levels[(x, y)] += 1
    discrepancy = sum(levels.values())
    return levels, discrepancy

def calculate_distances(lines, intersections):
    distances = []
    for intersection in intersections:
        for i, (line1, line2) in enumerate(combinations(lines, 2)):
            dist = np.linalg.norm(np.array(line1).mean(axis=0) - np.array(line2).mean(axis=0))
            distances.append((f'L{i+1}', f'L{i+2}', dist, line1, line2))
    return distances

def main():
    num_lines = int(input("Ingrese el número de líneas: "))
    line_length = float(input("Ingrese la longitud de las líneas (en metros): "))
    lines = []
    for i in range(num_lines):
        x1, y1, x2, y2 = np.random.rand(4) * line_length
        lines.append(((x1, y1), (x2, y2)))

    intersections = calculate_intersection(lines)
    draw_lines(lines, intersections)

    levels, discrepancy = calculate_levels_and_discrepancy(lines)
    print("Niveles:", levels)
    print("Discrepancia:", discrepancy)

    distances = calculate_distances(lines, intersections)
    print("Distancias entre líneas intersectadas:")
    for distance in distances:
        print(f"{distance[0]} vs {distance[1]}: {distance[2]:.2f} metros, L1: {distance[3]}, L2: {distance[4]}")

if __name__ == "__main__":
    main()
```

## Descripción:Este código genera un conjunto de líneas aleatorias en un plano 2D y luego detecta las intersecciones entre ellas. Posteriormente, calcula las distancias desde los vértices de intersección hacia los extremos de las líneas. Finalmente, muestra


```python
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
import itertools

def draw_lines(lines, intersections):
    plt.figure(figsize=(8, 8))
    colors = plt.cm.rainbow(np.linspace(0, 1, len(lines)))  # Genera colores únicos
    for i, line in enumerate(lines):
        (x1, y1), (x2, y2) = line
        plt.plot([x1, x2], [y1, y2], color=colors[i], marker='o', label=f'Línea {i+1}')
    for intersection in intersections:
        plt.plot(intersection[0], intersection[1], 'ro')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Líneas y Puntos de Intersección')
    plt.legend()
    plt.grid(True)
    plt.show()

def calculate_intersection(lines):
    intersections = []
    for i, line1 in enumerate(lines):
        for line2 in lines[i+1:]:
            xdiff = (line1[0][0] - line1[1][0], line2[0][0] - line2[1][0])
            ydiff = (line1[0][1] - line1[1][1], line2[0][1] - line2[1][1])

            def det(a, b):
                return a[0] * b[1] - a[1] * b[0]

            div = det(xdiff, ydiff)
            if div == 0:
                continue
            d = (det(*line1), det(*line2))
            x = det(d, xdiff) / div
            y = det(d, ydiff) / div
            if min(line1[0][0], line1[1][0]) <= x <= max(line1[0][0], line1[1][0]) and min(line1[0][1], line1[1][1]) <= y <= max(line1[0][1], line1[1][1]):
                if min(line2[0][0], line2[1][0]) <= x <= max(line2[0][0], line2[1][0]) and min(line2[0][1], line2[1][1]) <= y <= max(line2[0][1], line2[1][1]):
                    intersections.append((x, y))
    return intersections

def calculate_vertex_distances(lines, intersections):
    vertex_distances = defaultdict(list)
    for inter1, inter2 in itertools.combinations(intersections, 2):
        for line in lines:
            (x1, y1), (x2, y2) = line
            dist1 = np.linalg.norm(np.array(inter1) - np.array((x1, y1)))
            dist2 = np.linalg.norm(np.array(inter1) - np.array((x2, y2)))
            if dist1 + dist2 == np.linalg.norm(np.array((x1, y1)) - np.array((x2, y2))):  # Verifica si el punto inter1 está en la línea
                vertex_distances[(inter1, inter2)].append(dist1)
        for line in lines:
            (x1, y1), (x2, y2) = line
            dist1 = np.linalg.norm(np.array(inter2) - np.array((x1, y1)))
            dist2 = np.linalg.norm(np.array(inter2) - np.array((x2, y2)))
            if dist1 + dist2 == np.linalg.norm(np.array((x1, y1)) - np.array((x2, y2))):  # Verifica si el punto inter2 está en la línea
                vertex_distances[(inter1, inter2)].append(dist1)
    return vertex_distances

def main():
    num_lines = int(input("Ingrese el número de líneas aleatorias a generar: "))
    lines = []
    line_lengths = []
    for i in range(num_lines):
        length = float(input(f"Ingrese la longitud en metros de la línea {i+1}: "))
        line_lengths.append(length)
        x1, y1, x2, y2 = np.random.rand(4) * 10
        lines.append(((x1, y1), (x2, y2)))

    intersections = calculate_intersection(lines)
    draw_lines(lines, intersections)

    vertex_distances = calculate_vertex_distances(lines, intersections)
    for i, ((inter1, inter2), distances) in enumerate(vertex_distances.items(), 1):
        print(f"Distancias {i}: {distances} metros entre {inter1} y {inter2}")

if __name__ == "__main__":
    main()
```

# Detector de Intersecciones entre Líneas en Espacio Tridimensional

## Descripción:Este script genera líneas aleatorias en un espacio tridimensional y luego detecta las intersecciones entre estas líneas. Las líneas se representan como segmentos con puntos finales marcados en un gráfico tridimensional utilizando Matplotlib.

##El proceso de detección de intersecciones se realiza mediante la comparación de cada par de líneas generadas. Si dos líneas se cruzan en algún punto del espacio tridimensional, se registra la coordenada de la intersección y se indica qué líneas están involucradas en la intersección.

## Requisitos del Usuario:
- Especificar el número de líneas a generar (hasta un máximo de 99).


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def generate_lines(num_lines):
    lines = []
    for i in range(num_lines):
        x1, y1, z1 = np.random.rand(3) * 10
        x2, y2, z2 = np.random.rand(3) * 10
        lines.append(((x1, y1, z1), (x2, y2, z2)))
    return lines

def plot_lines(lines):
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    for line in lines:
        (x1, y1, z1), (x2, y2, z2) = line
        ax.plot([x1, x2], [y1, y2], [z1, z2], marker='o')

    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Líneas en 3D')
    plt.show()

def calculate_intersections(lines):
    intersections = []
    for i, line1 in enumerate(lines):
        for line2 in lines[i+1:]:
            (x1, y1, z1), (x2, y2, z2) = line1
            (x3, y3, z3), (x4, y4, z4) = line2

            denominator = ((x1 - x2) * (y3 - y4) - (y1 - y2) * (x3 - x4)) ** 2
            if denominator == 0:
                continue

            x = (((x1 * y2 - y1 * x2) * (x3 - x4) - (x1 - x2) * (x3 * y4 - y3 * x4)) / denominator) ** 2
            y = (((x1 * y2 - y1 * x2) * (y3 - y4) - (y1 - y2) * (x3 * y4 - y3 * x4)) / denominator) ** 2
            z = (((x1 - x2) * (y3 * z4 - z3 * y4) - (y1 - y2) * (x3 * z4 - z3 * x4)) / denominator) ** 2
            intersections.append(((x, y, z), i))
    return intersections

def main():
    num_lines = int(input("Ingrese el número de líneas a generar (máximo 99): "))
    if num_lines > 99:
        print("El número máximo de líneas es 99.")
        return

    lines = generate_lines(num_lines)
    plot_lines(lines)

    intersections = calculate_intersections(lines)
    print("Lista de datos:")
    for intersection in intersections:
        print(f"Número de vértices que cruzan en línea {intersection[1]+1}: {intersection[0]}")

if __name__ == "__main__":
    main()
```

# COMBINACIONES ALEATORIA EN  UN PLANO X,Y,Z

```python

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def generate_lines(num_lines):
    lines = []
    for i in range(num_lines):
        x1, y1, z1 = np.random.rand(3) * 10
        x2, y2, z2 = np.random.rand(3) * 10
        lines.append(((x1, y1, z1), (x2, y2, z2)))
    return lines

def generate_random_line_combinations(num_combinations, num_lines):
    combinations = []
    for _ in range(num_combinations):
        lines = generate_lines(num_lines)
        combinations.append(lines)
    return combinations

def plot_lines(lines):
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    for line in lines:
        (x1, y1, z1), (x2, y2, z2) = line
        ax.plot([x1, x2], [y1, y2], [z1, z2], marker='o')

    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Líneas en 3D')
    plt.show()

def calculate_intersections(lines):
    intersections = []
    for i, line1 in enumerate(lines):
        for line2 in lines[i+1:]:
            (x1, y1, z1), (x2, y2, z2) = line1
            (x3, y3, z3), (x4, y4, z4) = line2

            denominator = ((x1 - x2) * (y3 - y4) - (y1 - y2) * (x3 - x4)) ** 2
            if denominator == 0:
                continue

            x = (((x1 * y2 - y1 * x2) * (x3 - x4) - (x1 - x2) * (x3 * y4 - y3 * x4)) / denominator) ** 2
            y = (((x1 * y2 - y1 * x2) * (y3 - y4) - (y1 - y2) * (x3 * y4 - y3 * x4)) / denominator) ** 2
            z = (((x1 - x2) * (y3 * z4 - z3 * y4) - (y1 - y2) * (x3 * z4 - z3 * x4)) / denominator) ** 2
            intersections.append(((x, y, z), i))
    return intersections

def main():
    num_lines = int(input("Ingrese el número de líneas a generar (máximo 99): "))
    combinations = generate_random_line_combinations(30, num_lines)
    for i, lines in enumerate(combinations):
        print(f"Combinación {i+1}:")
        plot_lines(lines)
        intersections = calculate_intersections(lines)
        print("Número de intersecciones:", len(intersections))
        print()

if __name__ == "__main__":
    main()
```
